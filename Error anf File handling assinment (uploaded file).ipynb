### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
file = open("numbers.txt", "r")

total = 0

for line in file:
    number = int(line)
    total = total + number

print(total)

file.close()
```

### Q6 
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
# Q6 solution code here
file = open("numbers.txt", "r")

count = 0

for line in file:
    count = count + 1

print("Total lines:", count)

file.close()
```

### Q11

### Q13 (Easy)

### Q14

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print("Result:", result)

except (ZeroDivisionError, ValueError):
    print("Error: Please enter a valid non-zero number.")
```

### Q18

```python
# Q18 solution code here
try:
    number = 10 / 2
    print("Result:", number)

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

else:
    print("No error occurred.")
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    print("Age is valid.")


try:
    validate_age(16)
except ValueError as e:
    print(e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
file = None

try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("Error: File not found.")

finally:
    if file is not None:
        file.close()
        print("File closed.")
```
